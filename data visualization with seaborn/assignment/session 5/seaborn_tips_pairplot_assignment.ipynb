**Tasks**
1. Load the `tips` dataset and create a basic pairplot for all numerical columns.

4. Use `vars` to visualize two numerical columns with `hue='time'`.


## Import Libraries

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset from Seaborn
tips = sns.load_dataset("tips")

tips.head()
```

## Task 1 — Basic Pairplot for All Numerical Columns

A basic pairplot is created using all numerical columns from the `tips` dataset.


```python
sns.pairplot(tips.select_dtypes(include="number"))
plt.show()
```

## Task 2 — Pairplot with `hue='sex'`

The `sex` column is used as the hue so that observations are colored according to gender.


```python
sns.pairplot(tips, hue="sex")
plt.show()
```

## Task 3 — Pairplot with `hue='smoker'` and `diag_kind='kde'`

Kernel density estimates are displayed on the diagonal using `diag_kind='kde'`.


```python
sns.pairplot(tips, hue="smoker", diag_kind="kde")
plt.show()
```

## Task 4 — Select Two Numerical Columns Using `vars`

Here, `total_bill` and `tip` are selected with the required `vars` parameter, and the points are colored by `time` (Lunch/Dinner).


```python
sns.pairplot(
    tips,
    vars=["total_bill", "tip"],
    hue="time"
)
plt.show()
```
