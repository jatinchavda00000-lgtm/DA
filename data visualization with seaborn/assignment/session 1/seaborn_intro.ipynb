## Task 1 — Import libraries and display versions

```python
import seaborn as sns
import matplotlib.pyplot as plt

print("Seaborn version:", sns.__version__)
print("Matplotlib version:", plt.matplotlib.__version__)
```

## Task 2 — Weekly temperature trend using `sns.lineplot()`

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.set_theme(style='darkgrid')
fig, ax = plt.subplots(figsize=(8, 5))
sns.lineplot(x=days, y=temps, marker='o', ax=ax)
ax.set_xlabel('Day')
ax.set_ylabel('Temperature (°C)')
ax.set_title('Weekly Temperature Trend')
plt.tight_layout()
plt.show()
```

## Task 3 — Axes-level vs figure-level functions

The axes-level version uses `sns.lineplot()` and draws directly onto a Matplotlib `Axes` object. The figure-level version uses `sns.relplot()`, which creates and manages its own figure-level layout (including the axes), making it more convenient for faceting or multiple panels.

```python
# Axes-level function: sns.lineplot()
sns.set_theme(style='darkgrid')
fig, ax = plt.subplots(figsize=(8, 5))
sns.lineplot(x=days, y=temps, marker='o', ax=ax)
ax.set_xlabel('Day')
ax.set_ylabel('Temperature (°C)')
ax.set_title('Weekly Temperature Trend — Axes-Level')
plt.tight_layout()
plt.show()
```

```python
# Figure-level function: sns.relplot()
sns.set_theme(style='darkgrid')
g = sns.relplot(x=days, y=temps, kind='line', marker='o', height=5, aspect=1.6)
g.set_axis_labels('Day', 'Temperature (°C)')
g.set(title='Weekly Temperature Trend — Figure-Level')
plt.show()
```

### Observed difference

- **Axes-level (`sns.lineplot`)**: works directly with a specific Matplotlib axes, giving more granular control over where the plot is drawn.
- **Figure-level (`sns.relplot`)**: creates the figure and axes for you and is especially useful when building faceted/multi-panel relational plots.

## Task 4 — Compare Seaborn themes and save three separate images

```python
from pathlib import Path

output_dir = Path('seaborn_theme_images')
output_dir.mkdir(exist_ok=True)

styles = ['darkgrid', 'whitegrid', 'ticks']

for style in styles:
    sns.set_theme(style=style)
    fig, ax = plt.subplots(figsize=(8, 5))
    sns.lineplot(x=days, y=temps, marker='o', ax=ax)
    ax.set_xlabel('Day')
    ax.set_ylabel('Temperature (°C)')
    ax.set_title(f'Weekly Temperature Trend — {style}')
    fig.tight_layout()
    fig.savefig(output_dir / f'weekly_temperature_{style}.png', dpi=150, bbox_inches='tight')
    plt.close(fig)

print('Saved images:')
for style in styles:
    print(output_dir / f'weekly_temperature_{style}.png')
```

### Theme comparison

| Theme | Appearance |
|---|---|
| **darkgrid** | Uses a gray-toned background with visible grid lines, making it easy to read values across the plot. |
| **whitegrid** | Keeps a clean white background while retaining grid lines for easier visual alignment. |
| **ticks** | Uses a very clean background with tick marks but no prominent grid, giving a minimal look. |

**Overall:** `darkgrid` is the most grid-focused, `whitegrid` is cleaner while preserving guides, and `ticks` is the most minimal.
