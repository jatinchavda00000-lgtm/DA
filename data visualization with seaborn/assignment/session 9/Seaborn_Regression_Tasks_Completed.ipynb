# Seaborn Regression Plot Tasks

This notebook completes all 5 requested tasks using sample data and Seaborn regression plots.

**Libraries:** Pandas, NumPy, Matplotlib, Seaborn

```python
import os
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

OUTPUT_DIR = 'seaborn_task_plots'
os.makedirs(OUTPUT_DIR, exist_ok=True)

sns.set_theme(style='whitegrid')
print('Seaborn version:', sns.__version__)
```

## Task 1 — Instagram Likes vs Comments

Create a small DataFrame with at least 10 sample Instagram posts and visualize the relationship between likes and comments using `sns.regplot()`.

```python
instagram_df = pd.DataFrame({
    'post_id': range(1, 13),
    'likes': [120, 180, 250, 310, 420, 510, 610, 730, 850, 990, 1150, 1320],
    'comments': [8, 12, 18, 21, 29, 34, 43, 51, 63, 72, 86, 101]
})

print(instagram_df)

plt.figure(figsize=(8, 5))
sns.regplot(data=instagram_df, x='likes', y='comments', marker='o', scatter_kws={'s': 55}, line_kws={'linewidth': 2})
plt.title('Instagram Likes vs Comments')
plt.xlabel('Number of Likes')
plt.ylabel('Number of Comments')
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, 'task1_instagram_regplot.png'), dpi=150, bbox_inches='tight')
plt.show()
```

## Task 2 — Zomato-Style Price vs Delivery Time

Create a minimum 15-row sample dataset and use `sns.lmplot()` to show the regression relationship between order price and delivery time.

```python
zomato_df = pd.DataFrame({
    'order_id': range(1, 21),
    'order_price': [120, 180, 250, 150, 320, 450, 210, 280, 390, 520, 175, 230, 340, 410, 275, 360, 490, 200, 305, 440],
    'delivery_time_min': [24, 27, 31, 25, 36, 45, 29, 34, 40, 52, 28, 30, 37, 43, 33, 39, 48, 29, 35, 46]
})

print(zomato_df)

plot = sns.lmplot(
    data=zomato_df,
    x='order_price',
    y='delivery_time_min',
    height=5,
    aspect=1.45,
    scatter_kws={'s': 55},
    line_kws={'linewidth': 2}
)
plot.set_axis_labels('Food Order Price (₹)', 'Delivery Time (minutes)')
plot.fig.suptitle('Zomato-Style Order Price vs Delivery Time', y=1.03)
plot.savefig(os.path.join(OUTPUT_DIR, 'task2_zomato_lmplot.png'), dpi=150, bbox_inches='tight')
plt.show()
```

## Task 3 — Spotify Playlist Songs vs Total Duration with LOWESS

Use `lowess=True` in `sns.regplot()` to add a locally weighted smoothing curve. A second plot is included for comparison with the ordinary linear regression fit.

```python
spotify_df = pd.DataFrame({
    'songs': [8, 12, 16, 20, 24, 28, 32, 36, 40, 45, 50, 55, 60, 65, 70],
    'duration_minutes': [28, 41, 57, 68, 80, 94, 116, 129, 150, 162, 181, 205, 218, 245, 258]
})

print(spotify_df)

# LOWESS smoothing
plt.figure(figsize=(8, 5))
sns.regplot(
    data=spotify_df,
    x='songs',
    y='duration_minutes',
    lowess=True,
    scatter_kws={'s': 55}
)
plt.title('Spotify Playlist Songs vs Total Duration — LOWESS')
plt.xlabel('Number of Songs')
plt.ylabel('Total Playlist Duration (minutes)')
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, 'task3_spotify_lowess.png'), dpi=150, bbox_inches='tight')
plt.show()

# Comparison: ordinary linear regression
plt.figure(figsize=(8, 5))
sns.regplot(
    data=spotify_df,
    x='songs',
    y='duration_minutes',
    scatter_kws={'s': 55},
    line_kws={'linewidth': 2}
)
plt.title('Spotify Playlist Songs vs Total Duration — Linear Fit')
plt.xlabel('Number of Songs')
plt.ylabel('Total Playlist Duration (minutes)')
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, 'task3_spotify_linear_comparison.png'), dpi=150, bbox_inches='tight')
plt.show()
```

## Task 4 — Faceted Regression by Preparation Method

Use `lmplot(..., col='preparation_method')` to compare YouTube hours vs exam scores for two preparation methods: self-study and coaching.

```python
youtube_df = pd.DataFrame({
    'hours_youtube': [1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0, 1.0, 1.5, 2.0, 2.5, 3.0, 3.5, 4.0, 4.5, 5.0, 5.5, 6.0],
    'exam_score': [88, 84, 82, 79, 75, 72, 69, 65, 62, 76, 74, 72, 75, 78, 80, 82, 84, 87, 89, 91],
    'preparation_method': ['self-study'] * 9 + ['coaching'] * 11
})

print(youtube_df)

plot = sns.lmplot(
    data=youtube_df,
    x='hours_youtube',
    y='exam_score',
    col='preparation_method',
    height=4.5,
    aspect=1.05,
    ci=95,
    scatter_kws={'s': 55},
    line_kws={'linewidth': 2}
)
plot.set_axis_labels('Hours Spent on YouTube', 'Exam Score')
plot.set_titles('{col_name}')
plot.fig.suptitle('YouTube Usage vs Exam Score by Preparation Method', y=1.04)
plot.savefig(os.path.join(OUTPUT_DIR, 'task4_youtube_faceted_lmplot.png'), dpi=150, bbox_inches='tight')
plt.show()
```

## Task 5 — Flipkart Product Price vs Number of Reviews

The regression logic stays unchanged; only plot aesthetics are modified by adding a title, axis labels, and a custom marker color.

```python
flipkart_df = pd.DataFrame({
    'product_price': [499, 799, 999, 1299, 1599, 1999, 2499, 2999, 3499, 3999, 4499, 4999, 5999, 6999, 7999],
    'reviews': [2400, 2100, 1900, 1700, 1450, 1260, 1120, 980, 860, 760, 690, 610, 520, 430, 360]
})

print(flipkart_df)

# Same lmplot regression logic; only aesthetics are customized.
plot = sns.lmplot(
    data=flipkart_df,
    x='product_price',
    y='reviews',
    height=5,
    aspect=1.45,
    scatter_kws={'color': 'darkorange', 's': 60},
    line_kws={'linewidth': 2}
)
plot.set_axis_labels('Product Price (₹)', 'Number of Reviews')
plot.fig.suptitle('Flipkart Product Price vs Number of Reviews', y=1.03)
plot.savefig(os.path.join(OUTPUT_DIR, 'task5_flipkart_aesthetics.png'), dpi=150, bbox_inches='tight')
plt.show()
```
