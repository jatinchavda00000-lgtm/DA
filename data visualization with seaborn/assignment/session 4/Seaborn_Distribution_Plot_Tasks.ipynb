```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

## Task 1 — Flipkart Product Price Distribution

Load a CSV file containing Flipkart product prices and use `sns.boxplot()` to visualize the price distribution.


```python
# Load the sample Flipkart CSV
flipkart_df = pd.read_csv(r"/mnt/data/flipkart_zomato_ipl_fitness_tasks/flipkart_prices.csv")

print(flipkart_df.head())
print("\nSummary statistics:")
print(flipkart_df["price"].describe())
```

```python
plt.figure(figsize=(9, 4))
sns.boxplot(x=flipkart_df["price"])
plt.title("Flipkart Product Price Distribution")
plt.xlabel("Price (₹)")
plt.show()
```

**Interpretation:** The boxplot shows the median, middle 50% of product prices (the box), and the whiskers. Very expensive products may appear beyond the upper whisker, indicating unusually high prices compared with most products.


## Task 2 — Zomato Order Amounts by Restaurant Category

Compare order amounts for two restaurant categories: **Fast Food** and **Desserts**.


```python
# Load the sample Zomato dataset
zomato_df = pd.read_csv(r"/mnt/data/flipkart_zomato_ipl_fitness_tasks/zomato_orders.csv")

print(zomato_df.head())
print("\nAverage order amount by category:")
print(zomato_df.groupby("category")["order_amount"].mean())
```

```python
plt.figure(figsize=(8, 5))
sns.violinplot(
    data=zomato_df,
    x="category",
    y="order_amount",
    inner="box"
)
plt.title("Zomato Order Amount Distribution: Fast Food vs Desserts")
plt.xlabel("Restaurant Category")
plt.ylabel("Order Amount (₹)")
plt.show()
```

**Interpretation:** The width of each violin represents the density of order amounts. Wider sections mean more observations are concentrated around those amounts. Differences in the center and spread help compare typical spending and variability between the two categories.


## Task 3 — IPL Match Ticket Prices and Outliers

Create a boxplot from IPL ticket prices and identify outliers. Points outside the whiskers are treated as outliers.


```python
# Sample IPL ticket prices (₹)
ipl_prices = np.array([
    500, 600, 700, 750, 800, 850, 900, 950, 1000, 1100,
    1200, 1250, 1300, 1400, 1500, 1600, 1800, 2000, 2200, 2500,
    6000
])

plt.figure(figsize=(10, 4))
sns.boxplot(x=ipl_prices)
plt.title("IPL Match Ticket Price Distribution")
plt.xlabel("Ticket Price (₹)")
plt.show()

# Detect outliers using the standard 1.5 × IQR rule used for boxplot whiskers
q1 = np.percentile(ipl_prices, 25)
q3 = np.percentile(ipl_prices, 75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = ipl_prices[(ipl_prices < lower_bound) | (ipl_prices > upper_bound)]

print(f"Q1 = ₹{q1:.2f}")
print(f"Q3 = ₹{q3:.2f}")
print(f"IQR = ₹{iqr:.2f}")
print(f"Lower bound = ₹{lower_bound:.2f}")
print(f"Upper bound = ₹{upper_bound:.2f}")
print("Outliers:", outliers.tolist())
```

**Result:** The ticket price **₹6,000** is identified as an outlier because it lies above the upper whisker boundary calculated from the 1.5×IQR rule. This represents an unusually expensive ticket compared with the rest of the sample.


## Task 4 — Daily Step Counts from a Fitness App

Use `sns.violinplot()` to visualize daily steps and explain what the shape tells us about user activity.


```python
fitness_df = pd.read_csv(r"/mnt/data/flipkart_zomato_ipl_fitness_tasks/fitness_steps.csv")

print(fitness_df.head())
print("\nStep-count summary:")
print(fitness_df["daily_steps"].describe())
```

```python
plt.figure(figsize=(6, 6))
sns.violinplot(y=fitness_df["daily_steps"], inner="box")
plt.title("Daily Step Count Distribution")
plt.ylabel("Steps per Day")
plt.show()
```

**Interpretation of the violin shape:**

- **Wide sections** show step-count ranges where many days were recorded.
- **Narrow sections** show less common activity levels.
- A violin that is wider around higher step counts suggests the user was frequently more active.
- A wider lower section suggests many low-activity days.
- The overall height shows the range of daily step counts.

In this sample, the distribution represents variation in daily activity rather than a single fixed activity level.
