# Seaborn Distribution Tasks

This notebook completes all four requested tasks using Seaborn. The datasets are sample/synthetic datasets created for demonstration and assignment use.


```python
import seaborn as sns
import matplotlib.pyplot as plt

# Optional styling for clearer plots
sns.set_theme(style='whitegrid')
```

## Task 1 — Delivery Time Distribution

Use `sns.histplot()` to visualize delivery times (in minutes) for a sample Swiggy order dataset with more than 30 values.

```python
delivery_times = [24, 31, 28, 35, 42, 27, 33, 29, 45, 38, 32, 26, 41, 37, 30,
                  34, 39, 25, 28, 36, 44, 31, 29, 40, 35, 27, 33, 46, 38, 32,
                  30, 43, 36, 29, 34, 41]

plt.figure(figsize=(8, 5))
sns.histplot(delivery_times, bins=8, kde=False)
plt.title('Distribution of Swiggy Delivery Times')
plt.xlabel('Delivery Time (minutes)')
plt.ylabel('Number of Orders')
plt.show()
```

## Task 2 — Daily Step Count Distribution

A one-month (30-day) sample of daily step counts is created below. `sns.kdeplot()` is used to show the probability density curve.

```python
daily_steps = [6540, 8230, 7120, 9450, 11020, 5830, 7640, 8920, 10110, 6780,
               7350, 8120, 9650, 10430, 7210, 8560, 6930, 9780, 11540, 6420,
               7890, 9340, 8760, 10560, 7080, 8210, 9670, 11320, 7520, 8890]

plt.figure(figsize=(8, 5))
sns.kdeplot(daily_steps, fill=True, linewidth=2)
plt.title('KDE of Daily Step Counts (30 Days)')
plt.xlabel('Daily Steps')
plt.ylabel('Probability Density')
plt.show()
```

## Task 3 — Zomato Monthly Spending: Histogram + KDE

The same monthly spending dataset is used for both `sns.histplot()` and `sns.kdeplot()` on the **same axes**, as requested in the hint.

```python
zomato_spending = [180, 240, 320, 450, 275, 390, 520, 610, 340, 280, 470, 550,
                   690, 310, 225, 430, 365, 580, 750, 495, 260, 335, 410, 625,
                   710, 380, 455, 590, 670, 290]

fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(zomato_spending, bins=8, stat='density', alpha=0.5, ax=ax)
sns.kdeplot(zomato_spending, linewidth=2, ax=ax)
ax.set_title('Zomato Monthly Spending — Histogram with KDE')
ax.set_xlabel('Monthly Spending (₹)')
ax.set_ylabel('Density')
plt.show()
```

## Task 4 — IPL Match Score Distribution and Bin Comparison

The same IPL score dataset is plotted with 5, 10, and 20 bins to compare how the choice of bin count changes the histogram.

```python
ipl_scores = [145, 178, 132, 201, 165, 187, 154, 212, 149, 173,
              190, 128, 156, 181, 205, 142, 169, 195, 137, 221,
              160, 176, 148, 188, 214, 134, 159, 199, 171, 183]

for bins in [5, 10, 20]:
    plt.figure(figsize=(8, 4))
    sns.histplot(ipl_scores, bins=bins)
    plt.title(f'IPL Run Distribution (bins={bins})')
    plt.xlabel('Runs')
    plt.ylabel('Number of Matches')
    plt.show()
```

### Observation

With **5 bins**, the histogram is smoother and gives a broad overall view, but some detail is hidden. With **10 bins**, the distribution becomes more informative without being too noisy. With **20 bins**, the bars are narrower and reveal more small variations, but the plot can look more irregular because each bin contains fewer observations.

## Conclusion

Seaborn's `histplot()` is useful for showing frequency distributions, while `kdeplot()` provides a smooth estimate of probability density. Overlaying both on the same axes makes it easier to understand the shape of a spending distribution.
