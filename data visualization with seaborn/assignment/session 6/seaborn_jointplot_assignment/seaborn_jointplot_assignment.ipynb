# Seaborn Jointplot Assignment

This notebook completes all four requested tasks using Seaborn `jointplot`.

> **Note:** No source CSV files were provided with the question, so representative datasets were created with realistic columns matching the task descriptions. The CSV files are included in this folder and can be replaced with your own files without changing the plotting logic.


```python
# Install (only if needed)
# %pip install pandas numpy seaborn matplotlib

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
```

## Task 1 — Flipkart: Rating vs Price

Load the CSV file and use `sns.jointplot(..., kind='scatter')` to visualize the relationship between product rating and price.


```python
flipkart_df = pd.read_csv("flipkart_products.csv")
print(flipkart_df.head())
print("\nShape:", flipkart_df.shape)

sns.jointplot(
    x="rating",
    y="price",
    data=flipkart_df,
    kind="scatter",
    height=7
)
plt.suptitle("Flipkart Products: Rating vs Price", y=1.02)
plt.show()
```

**Interpretation:** The scatter jointplot shows individual products, while the marginal plots summarize the distributions of rating and price. In this representative dataset, price varies substantially across ratings, so there is no strong linear relationship.


## Task 2 — Swiggy: Delivery Time vs Order Value

Use a hexbin-style jointplot (`kind='hex'`) to identify dense regions of orders.


```python
swiggy_df = pd.read_csv("swiggy_orders.csv")
print(swiggy_df.head())
print("\nShape:", swiggy_df.shape)

sns.jointplot(
    x="delivery_time",
    y="order_value",
    data=swiggy_df,
    kind="hex",
    height=7
)
plt.suptitle("Swiggy Orders: Delivery Time vs Order Value", y=1.02)
plt.show()
```

**Interpretation:** The hex plot groups observations into hexagonal bins. Darker/more populated bins reveal where orders are concentrated, making it easier than a raw scatterplot to see dense regions when many points overlap.


## Task 3 — Instagram: Likes vs Comments

Use a KDE jointplot to visualize the concentration and shape of the two-variable distribution.


```python
instagram_df = pd.read_csv("instagram_posts.csv")
print(instagram_df.head())
print("\nShape:", instagram_df.shape)

sns.jointplot(
    x="likes",
    y="comments",
    data=instagram_df,
    kind="kde",
    fill=True,
    height=7
)
plt.suptitle("Instagram Posts: Likes vs Comments", y=1.02)
plt.show()
```

**Interpretation:** KDE smooths the observations into density contours. The densest area shows where likes and comments commonly occur together. This is helpful for identifying concentration and distribution shape rather than individual observations.


## Task 4 — Zomato: Compare Scatter, Hex, and KDE

The same Zomato dataset is visualized three ways so that the differences between plot types are easy to compare.


```python
zomato_df = pd.read_csv("zomato_reviews.csv")
print(zomato_df.head())
print("\nShape:", zomato_df.shape)

# 1) Scatter
sns.jointplot(
    x="rating",
    y="number_of_reviews",
    data=zomato_df,
    kind="scatter",
    height=7
)
plt.suptitle("Zomato Reviews — Scatter", y=1.02)
plt.show()

# 2) Hex
sns.jointplot(
    x="rating",
    y="number_of_reviews",
    data=zomato_df,
    kind="hex",
    height=7
)
plt.suptitle("Zomato Reviews — Hex", y=1.02)
plt.show()

# 3) KDE
sns.jointplot(
    x="rating",
    y="number_of_reviews",
    data=zomato_df,
    kind="kde",
    fill=True,
    height=7
)
plt.suptitle("Zomato Reviews — KDE", y=1.02)
plt.show()
```

```python
# Optional: numerical summary for the Zomato dataset
print("Correlation:", zomato_df["rating"].corr(zomato_df["number_of_reviews"]))
print("\nDescriptive statistics:")
display(zomato_df[["rating", "number_of_reviews"]].describe())
```

### Comparison and conclusion

**Scatter:** Best for seeing individual restaurants and spotting outliers. However, many points can overlap, hiding the densest regions.

**Hex:** Best for understanding where observations are concentrated when there are many restaurants. The binned structure makes high-density areas easier to see.

**KDE:** Best for understanding the smooth overall density and shape of the relationship. It gives a clean visual of where observations are most concentrated, but it does not show individual restaurants or exact bin counts.

**Best choice for this dataset: `hex`.** For a restaurant-review dataset with many observations and a strongly right-skewed review count, the hex plot gives the clearest view of concentration without the overplotting of scatter and the smoothing assumptions of KDE. Scatter is still useful for outliers, while KDE is useful for distribution shape.
